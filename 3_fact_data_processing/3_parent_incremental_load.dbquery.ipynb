```sql
DESCRIBE TABLE fmcg.gold.fact_orders

COPY INTO fmcg.gold.fact_orders
FROM (
  SELECT 
  CAST(date AS DATE) AS date,
  CAST(product_code AS STRING) AS product_code,
  CAST(customer_code AS BIGINT) AS customer_code,
  CAST(sold_quantity AS BIGINT) AS sold_quantity
  FROM '/Volumes/fmcg/gold/parent_incremental_update/fact_orders.csv'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = "true")


```
